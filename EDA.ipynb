## 7. Bivariate Analysis

Bivariate analysis examines the relationship between two variables. In this section, the relationship between the Happiness Score and several explanatory factors is explored using scatter plots and correlation coefficients.

### 7.1 GDP per Capita vs Happiness Score

```python
df["Explained by: GDP per capita"].describe()
```

```python
gdp_happiness_corr = np.corrcoef(
    df["Explained by: GDP per capita"],
    df["Happiness score"]
)[0, 1]

print("Correlation between GDP per capita and Happiness Score:",
      gdp_happiness_corr)
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df["Explained by: GDP per capita"],
    df["Happiness score"],
    alpha=0.7
)

plt.title("GDP per Capita vs Happiness Score")
plt.xlabel("GDP per Capita")
plt.ylabel("Happiness Score")

plt.show()
```

### Observation

The scatter plot shows the relationship between GDP per capita and Happiness Score. The correlation coefficient indicates the strength and direction of the linear relationship between the two variables.

A positive correlation suggests that countries with higher GDP per capita tend to have higher happiness scores, although correlation alone does not establish causation.

```python
x = df["Explained by: GDP per capita"]
y = df["Happiness score"]

slope, intercept = np.polyfit(x, y, 1)

trend_line = slope * x + intercept
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    x,
    y,
    alpha=0.7
)

plt.plot(
    x,
    trend_line
)

plt.title("GDP per Capita vs Happiness Score")
plt.xlabel("GDP per Capita")
plt.ylabel("Happiness Score")

plt.show()
```

## 8. Correlation Analysis

The correlation coefficients for the major explanatory factors are summarized below.

```python
correlation_summary = pd.DataFrame({
    "Factor": [
        "GDP per capita",
        "Social support",
        "Healthy life expectancy",
        "Freedom",
        "Generosity",
        "Perceptions of corruption"
    ],
    "Correlation with Happiness": [
        gdp_happiness_corr,
        social_support_corr,
        life_expectancy_corr,
        freedom_corr,
        generosity_corr,
        corruption_corr
    ]
})

correlation_summary
```

```python
correlation_summary.sort_values(
    by="Correlation with Happiness",
    ascending=False
)
```

```python
sorted_corr = correlation_summary.sort_values(
    by="Correlation with Happiness"
)

plt.figure(figsize=(10, 6))

plt.barh(
    sorted_corr["Factor"],
    sorted_corr["Correlation with Happiness"]
)

plt.title("Correlation of Factors with Happiness Score")
plt.xlabel("Correlation Coefficient")
plt.ylabel("Factor")

plt.tight_layout()
plt.show()
```

### Observation

The correlation analysis shows that the explanatory factors have different levels of association with the Happiness Score. Factors with higher positive correlation coefficients have a stronger positive linear relationship with happiness, while lower coefficients indicate a weaker relationship.

Correlation measures association and does not by itself prove that one factor causes changes in happiness.
